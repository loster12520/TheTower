## brainfuck kotlin 编译器

```kotlin
val sourceFilePath = "./source.brainfuck"
```

```kotlin
// 栈， 256 字节
val stack = List(256) { 0.toByte() }.toMutableList()

// 指针
var pointer = 0
```

```kotlin
import java.io.File

// 读取源代码
val sourceCode = File(sourceFilePath).readText()
```

```kotlin
// 解释执行
var pc = 0 // 程序计数器
while (pc < sourceCode.length) {
    when (sourceCode[pc]) {
        '>' -> pointer = (pointer + 1) % stack.size
        '<' -> pointer = (pointer - 1 + stack.size) % stack.size
        '+' -> stack[pointer] = (stack[pointer] + 1).toByte()
        '-' -> stack[pointer] = (stack[pointer] - 1).toByte()
        '.' -> print(stack[pointer].toInt().toChar())
        ',' -> {
            val input = System.`in`.read()
            stack[pointer] = if (input == -1) 0.toByte() else input.toByte()
        }
        '[' -> {
            if (stack[pointer] == 0.toByte()) {
                var bracketCount = 1
                while (bracketCount > 0) {
                    pc++
                    when (sourceCode[pc]) {
                        '[' -> bracketCount++
                        ']' -> bracketCount--
                    }
                }
            }
        }
        ']' -> {
            if (stack[pointer] != 0.toByte()) {
                var bracketCount = 1
                while (bracketCount > 0) {
                    pc--
                    when (sourceCode[pc]) {
                        ']' -> bracketCount++
                        '[' -> bracketCount--
                    }
                }
            }
        }
    }
    pc++
}
```
